### Data Feature Enginering 
create a numeric scale for the different question parttersn to convert responses to numeric values


```python
df_clean["ReceivedGift_Flag"] = df_clean["Q23_ReceivedGifts"].map({
    "Yes": 2, "Maybe": 1, "No": 0
})

df_clean["Target"] = df_clean["Target"].map({
    "Yes": 2, "Maybe": 1, "No": 0
})
```

```python
df_clean.head()
```

```python
df_clean = df_clean.drop(columns=["OfferedMalpractice_Flag", "ReceivedGift_Flag"])
df_clean.head(10)
```

```python
# Map Likert scale to numeric 1–5 for all Likert-based columns
likert_map = {
    "Strongly disagree": 1,
    "Disagree": 2,
    "Neutral": 3,
    "Agree": 4,
    "Strongly agree": 5
}

likert_cols = [col for col in df_clean.columns if col.startswith("Q") and col not in [ "Q23_ReceivedGifts"]]

for col in likert_cols:
    df_clean[col] = df_clean[col].map(likert_map)
```
