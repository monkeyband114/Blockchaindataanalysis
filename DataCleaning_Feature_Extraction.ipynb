## Numerize Categorical Columns

##### One-Hot Encoding

Best for categories with no order, such as: Gender, Zone, EmploymentStatus, and Ordinal Encoding, which works for: Education level, Income bracket and Age groups

```python
ordinal_categories = {
    "AgeGroup": ["18 - 25", "26 - 35", "36 - 46", "46 - 60", "60+"],
    "HighestEducation": [
        "Did not attend", "Primary 6", "Secondary school certificate",
        "OND", "B.Sc/HND/B.Eng", "Post Graduate"
    ],
    "MonthlyPersonalIncome": [
        "Less than ₦20,000", "₦20,000 - ₦49,000", "₦50,000 - ₦99,000",
        "₦100,000 - ₦199,000", "₦200,000 - ₦499,000", "Above ₦500,000"
    ]
}

ordinal_cols = list(ordinal_categories.keys())

onehot_cols = [
    "Gender",
    "Zone",
    "EmploymentStatus"
]
```

```python
transformer = ColumnTransformer(
    transformers=[
        ("ordinal", OrdinalEncoder(categories=list(ordinal_categories.values())), ordinal_cols),
        ("onehot", OneHotEncoder(drop="first"), onehot_cols),
    ],
    remainder="passthrough"   # Keep every other column untouched
)
```

```python
transformed = transformer.fit_transform(df_clean)
```

```python
new_ordinal_cols = ordinal_cols
```

```python
# One hot encoded column names
onehot_feature_names = transformer.named_transformers_["onehot"].get_feature_names_out(onehot_cols)
```

```python
# Remainder columns
remainder_cols = [c for c in df_clean.columns if c not in ordinal_cols + onehot_cols]
```

```python
# Combine all new column names
all_columns = (
    new_ordinal_cols +
    list(onehot_feature_names) +
    remainder_cols
)
```

```python
df_encoded = pd.DataFrame(transformed, columns=all_columns)
```
